```python
import duckdb
import pandas as pd

pd.set_option('display.max_rows', 1000)    # ou um número maior

# Conecta ao banco já criado
con = duckdb.connect("../db_simulado.duckdb")
```

```python
df_tables = con.execute("SHOW TABLES").df()
print(df_tables)
```

```python
tabelas = df_tables['name'].tolist()

for tab in tabelas:
    df = con.execute(f"SELECT * FROM {tab}").df()
    print(f"\n===== {tab} =====")
    print(df.head(2))
    if tab == "Pedido":
        print("Data mais antiga: ", df.order_date.min())    
    print("Shape:", df.shape)
```

```python
# Listar todas as tabelas
print(con.execute("SHOW TABLES").fetchall())

```
